```python
# task1.py -G4_DV_L08_[Name]_Integration
# Grade 4 Dataville - Lesson 08
# Three- District Comparison Tool

# Purpose: Accept district data from user, produce comparison report
'''
Collects name, population (string), and data quality score (string)
for three Dataville districts. Produces a comparison report.
Note: Population and quality stay as strings (conversion in Unit 2).
'''

print('=' * 48)
print('DATAVILLE THREE-DISTRICT COMPARISON TOOL')
print('=' * 48)
print()

# INPUT - District 1
print('--- District 1---')
d1_name = input('Name: ')
d1_pop = input('Population: ')
d1_quality = input('Data quality(1-10): ')
d1_industry = input(' Main industry: ')
print()

#INPUT - District 2
print('--- District 2---')
d2_name = input('Name: ')
d2_pop = input('Population: ')
d2_quality = input('Data quality(1-10): ')
d2_industry = input('Main industry: ')
print()

# INPUT - District 3
print('---District 3---')
d3_name = input('Name: ')
d3_pop = input('Population: ')
d3_quality = input('Data quality: ')
d3_industry = input('Main industry: ')
print()

# OUTPUT - Comparison report
sep_heavy = '=' * 48
sep_light = '=' * 48

print()
print(sep_heavy)
print('THREE-DIUSTRICT COMPARISON REPORT')
print('Prepared by: Junior Analyst for Dr Amara Osei')
print(sep_heavy)

print(f'District: {d1_name}')
print(f'Population: {d1_pop}')
print(f'Quality: {int(d1_quality)/10}')
print(f'Industry: {d1_industry}')
print(sep_light)

print(f'District: {d2_name}')
print(f'Population: {d2_pop}')
print(f'Quality: {int(d2_quality)/10}')
print(f'Industry: {d2_industry}')
print(sep_light)

print(f'District: {d3_name}')
print(f'Population: {d3_pop}')
print(f'Quality: {int(d3_quality)/10}')
print(f'Industry: {d3_industry}')

print(sep_heavy)
print('Report complete. Ready for dashboard submission.')
```
